```python
import os
import pandas as pd
import cv2
import imageio
import numpy as np
import matplotlib.pyplot as plt
import io
from PIL import Image

# Define colors for correct (green) and incorrect (red) frames
GREEN = (0, 255, 0)
RED = (0, 0, 255)
FRAME_THICKNESS = 10  # Thickness of the frame
RIS = 1
SHOT = 16534

# Paths
csv_path = '/compass/Shared/Users/bogdanov/ml_tokamak/runs/24-05-27, 20-00-44 both, finer lr_scheduler, 3 classes, resnet34_all_layers/prediction_df.csv'
image_folder = "/compass/Shared/Users/bogdanov/ml_tokamak/imgs"

# Load results CSV
df = pd.read_csv(csv_path)
df = df[df["shot"] == SHOT]
# Sort the DataFrame by time and keep only the unique time values
df = df.sort_values(by='time').drop_duplicates(subset='time')

# Define label names
label_names = {0: "L-mode", 1: "H-mode", 2: "ELM"}

# Add label names and confidence to DataFrame
df["label_name"] = df["label"].map(label_names)
df["prediction_name"] = df["prediction"].map(label_names)
df["confidence"] = df.apply(lambda row: row[f"prob_{row['prediction']}"] if row["prediction"] in [0, 1, 2] else None, axis=1)


# Read the h_alpha signal data
df_150 = pd.read_csv(f'/compass/Shared/Users/bogdanov/ml_tokamak/data/h_alpha_signal_150kHz/shot_{SHOT}.csv')

# Crop the data to the interesting parts
df = df[(df['time']>=1120) & (df['time']<=1160)]
df_150 = df_150[(df_150['time']>=df['time'].min())&(df_150['time']<=df['time'].max())]

# Process images and create frames

frames = []
def new_func(df_150, time_val):
    plt.figure(figsize=(10, 3))  # Make plot wider
    plt.plot(df_150['time'], df_150['h_alpha'], label='h_alpha', linewidth=1.5)
    plt.axvline(x=time_val, color='r', linestyle='--', linewidth=2, label='Current time')
    plt.xlabel("Time [ms]", fontsize=14)
    plt.ylabel("h_alpha [-]", fontsize=14)
    plt.tick_params(axis='both', which='major', labelsize=13)
    plt.legend(fontsize=13)
    plt.grid(True)
    plt.tight_layout()

for _, row in df.iterrows():
    shot = int(row["shot"])
    prediction = row["prediction"]
    label = row["label"]
    time_val = row["time"]
    confidence = row["confidence"]
    prediction_name = row["prediction_name"]
    label_name = row["label_name"]

    # Determine frame color
    color = GREEN if prediction == label else RED

    # Construct image filename
    img_folder_path = os.path.join(image_folder, str(shot))
    img_filename = f"RIS{RIS}_{shot}_t={time_val:.1f}.png"
    img_path = os.path.join(img_folder_path, img_filename)

    # Load image
    if not os.path.exists(img_path):
        print(f"Warning: Image {img_path} not found. Skipping.")
        continue

    img = cv2.imread(img_path)
    if img is None:
        print(f"Warning: Could not read {img_path}. Skipping.")
        continue

    # Add border (frame) to image
    img = cv2.copyMakeBorder(img, FRAME_THICKNESS, FRAME_THICKNESS, FRAME_THICKNESS, FRAME_THICKNESS,
                             cv2.BORDER_CONSTANT, value=color)
    
    # Overlay text with prediction, ground truth, and confidence
    text = f"Pred: {prediction_name} ({confidence:.4f}) | GT: {label_name}"
    cv2.putText(img, text, (20, 40), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2, cv2.LINE_AA)
    
    # Convert to RGB (OpenCV loads images in BGR format)
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img_height, img_width = img_rgb.shape[:2]
    
    # Create the h_alpha plot with wider width
    new_func(df_150, time_val)

    # Save the plot to a buffer
    buf = io.BytesIO()
    plt.savefig(buf, format='png', dpi=100, bbox_inches='tight')
    buf.seek(0)
    plot_img = np.array(Image.open(buf))
    plt.close()

    # Convert the plot to RGB
    plot_img_rgb = cv2.cvtColor(plot_img, cv2.COLOR_RGBA2RGB)
    plot_height, plot_width = plot_img_rgb.shape[:2]
    
    # Use the plot's width as the target width
    target_width = plot_width
    
    # Create white padding to match the plot width
    pad_width = target_width - img_width
    if pad_width > 0:
        # Calculate left and right padding to center the image
        left_pad = pad_width // 2
        right_pad = pad_width - left_pad
        img_padded = cv2.copyMakeBorder(img_rgb, 0, 0, left_pad, right_pad, 
                                        cv2.BORDER_CONSTANT, value=(255, 255, 255))
    else:
        # If image is somehow wider than plot, resize the plot
        img_padded = img_rgb
        plot_img_rgb = cv2.resize(plot_img_rgb, (img_width, plot_height))
    
    # Resize the plot to maintain its aspect ratio while matching the padded image width
    plot_img_resized = cv2.resize(plot_img_rgb, (img_padded.shape[1], 
                                  int(plot_height * (img_padded.shape[1] / plot_width))))
    
    # Combine the padded image and the plot vertically
    combined_frame = np.concatenate((img_padded, plot_img_resized), axis=0)
    
    # Append to frames
    frames.append(combined_frame)


# Create output directory if it doesn't exist
output_dir = f"tokamak_frames_shot_{SHOT}"
os.makedirs(output_dir, exist_ok=True)
```

```python

# Save individual frames as PNG images
if frames:
    for i, frame in enumerate(frames):
        # Get the time value for this frame from the dataframe
        time_val = df.iloc[i]['time']
        # Save the frame as an image
        output_path = os.path.join(output_dir, f"frame_{i:03d}.png")
        cv2.imwrite(output_path, cv2.cvtColor(frame, cv2.COLOR_RGB2BGR))
    
    print(f"Saved {len(frames)} frames to {output_dir}")
else:
    print("No images processed. Check image paths and filenames.")
```

```python
from sklearn.metrics import f1_score, precision_score, recall_score, cohen_kappa_score
import numpy as np
from sklearn.metrics import confusion_matrix
import seaborn as sns

# Create a dataframe to store results
metrics_df = pd.DataFrame(columns=['metric', 'macro', 'weighted', 'class_0', 'class_1', 'class_2'])

# Calculate F1 score
f1_macro = f1_score(predictions['label'], predictions['prediction'], average='macro')
f1_weighted = f1_score(predictions['label'], predictions['prediction'], average='weighted')
f1_per_class = f1_score(predictions['label'], predictions['prediction'], average=None)

metrics_df.loc[0] = ['f1_score', f1_macro, f1_weighted, f1_per_class[0], f1_per_class[1], f1_per_class[2]]

# Calculate precision
precision_macro = precision_score(predictions['label'], predictions['prediction'], average='macro')
precision_weighted = precision_score(predictions['label'], predictions['prediction'], average='weighted')
precision_per_class = precision_score(predictions['label'], predictions['prediction'], average=None)

metrics_df.loc[1] = ['precision', precision_macro, precision_weighted, precision_per_class[0], precision_per_class[1], precision_per_class[2]]

# Calculate recall
recall_macro = recall_score(predictions['label'], predictions['prediction'], average='macro')
recall_weighted = recall_score(predictions['label'], predictions['prediction'], average='weighted')
recall_per_class = recall_score(predictions['label'], predictions['prediction'], average=None)

metrics_df.loc[2] = ['recall', recall_macro, recall_weighted, recall_per_class[0], recall_per_class[1], recall_per_class[2]]

# Calculate Cohen's kappa
kappa = cohen_kappa_score(predictions['label'], predictions['prediction'])

metrics_df.loc[3] = ['cohen_kappa', kappa, kappa, np.nan, np.nan, np.nan]

# Print the results
print(metrics_df)

# Calculate confusion matrix
import matplotlib.pyplot as plt

cm = confusion_matrix(predictions['label'], predictions['prediction'], normalize='true')
plt.figure(figsize=(10, 8))

# Format the annotations to show percentages with 1 decimal place
cm_percent = cm * 100
annotations = [[f'{val:.1f}%' for val in row] for row in cm_percent]

sns.heatmap(cm, annot=annotations, fmt='', cmap='Blues', 
            xticklabels=['L-mode', 'H-mode', 'ELM'],
            yticklabels=['L-mode', 'H-mode', 'ELM'],
            annot_kws={"size": 20})
plt.xlabel('Predicted Label', fontsize=24)
plt.ylabel('True Label', fontsize=24)
plt.title('Confusion Matrix', fontsize=24)
plt.xticks(fontsize=22)
plt.yticks(fontsize=22)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.patches as mpatches


## RESULTS

predictions = pd.read_csv('/compass/Shared/Users/bogdanov/ml_tokamak/runs/24-05-27, 20-00-44 both, finer lr_scheduler, 3 classes, resnet34_all_layers/prediction_df.csv', index_col=0)

ris_1_pred = predictions


def draw_pred_plot(shot, time_span = None):
    # Get data for this shot
    ris_1_pred_shot = ris_1_pred[ris_1_pred['shot']==shot]
    
    # Check if shot exists in the dataset
    if ris_1_pred_shot.empty:
        print(f"Warning: Shot {shot} not found in the dataset")
        return
    
    # Get h_alpha data for the shot
    h_alpha = pd.read_csv(f'/compass/Shared/Users/bogdanov/ml_tokamak/data/h_alpha_signal_150kHz/shot_{shot}.csv')
    
    # Merge the prediction data with h_alpha data
    joined = pd.merge_asof(ris_1_pred_shot, h_alpha, on='time', tolerance=1/150, direction='nearest')
    
    if time_span:
        joined = joined[(joined['time']>=time_span[0]) & (joined['time']<=time_span[1])]
    
    # Check if we have any data after filtering
    if joined.empty:
        print(f"Warning: No data found for shot {shot} in time range {time_span}")
        return
    
    # Reset index to avoid index issues
    joined = joined.reset_index(drop=True)
    
    # Create the figure with two subplots
    fig, axs = plt.subplots(2, 1, figsize=(12, 8), gridspec_kw={'height_ratios': [3, 1]}, sharex=True)
    
    # Top plot - H-alpha signal with colored background for predictions
    axs[0].plot(joined['time'], joined['h_alpha'], 'k-', lw=1.5, label='H-alpha signal')
    
    # Get min and max values for the colored regions
    y_min, y_max = axs[0].get_ylim()
    y_range = y_max - y_min
    y_min -= y_range * 0.1  # Add some padding
    y_max += y_range * 0.1
    axs[0].set_ylim(y_min, y_max)
    
    # Create a colormap for different modes
    modes = {0: 'L-mode', 1: 'H-mode', 2: 'ELM'}
    colors = {0: 'blue', 1: 'green', 2: 'red'}
    
    # Determine where predictions change
    joined['prediction_change'] = joined['prediction'] != joined['prediction'].shift(1)
    
    # Find change points for predictions
    pred_change_points = joined[joined['prediction_change']].index.tolist()
    
    # Add the first and last points to ensure we cover the entire range
    if len(pred_change_points) == 0 or pred_change_points[0] != 0:
        pred_change_points = [0] + pred_change_points
    if pred_change_points[-1] != len(joined) - 1:
        pred_change_points.append(len(joined) - 1)
    
    # Create the colored spans for predictions
    for i in range(len(pred_change_points) - 1):
        start_idx = pred_change_points[i]
        end_idx = pred_change_points[i+1]
        
        # Get the time values for this span
        start_time = joined.loc[start_idx, 'time']
        end_time = joined.loc[end_idx, 'time']
        
        # Get the prediction for this span
        prediction = joined.loc[start_idx, 'prediction']
        pred_color = colors[prediction]
        
        # Color based on prediction
        axs[0].axvspan(start_time, end_time, color=pred_color, alpha=0.3)
    
    # Bottom plot - show the actual mode/label
    # Plot the actual mode as colored spans on the bottom subplot
    mode_changes = joined['label'] != joined['label'].shift(1)
    mode_change_points = joined[mode_changes].index.tolist()
    
    # Add first and last points
    if len(mode_change_points) == 0 or mode_change_points[0] != 0:
        mode_change_points = [0] + mode_change_points
    if mode_change_points[-1] != len(joined) - 1:
        mode_change_points.append(len(joined) - 1)
    
    # Fill the bottom plot with mode information
    for i in range(len(mode_change_points) - 1):
        start_idx = mode_change_points[i]
        end_idx = mode_change_points[i+1]
        
        # Get the time values for this span
        start_time = joined.loc[start_idx, 'time']
        end_time = joined.loc[end_idx, 'time']
        
        # Get the mode for this span
        mode = joined.loc[start_idx, 'label']
        mode_name = modes[mode]
        mode_color = colors[mode]
        
        # Add colored span in the bottom subplot
        axs[1].axvspan(start_time, end_time, color=mode_color, alpha=0.3)
    
    # Add legend entries for the colored regions
    l_patch = mpatches.Patch(color=colors[0], alpha=0.3, label='L-mode')
    h_patch = mpatches.Patch(color=colors[1], alpha=0.3, label='H-mode')
    elm_patch = mpatches.Patch(color=colors[2], alpha=0.3, label='ELM')
    
    # Combine legend items and add to top plot
    handles = [l_patch, h_patch, elm_patch]
    handles.append(plt.Line2D([0], [0], color='k', lw=1.5, label='H-alpha signal'))
    axs[0].legend(handles=handles, loc='upper right')
    
    # Set labels and title
    axs[0].set_ylabel('H-alpha signal', fontsize=12)
    axs[0].set_title(f'H-alpha signal with model predictions for shot {shot}', fontsize=14)
    axs[0].grid(True, alpha=0.3)
    
    axs[1].set_yticks([])  # Hide y-axis ticks in bottom plot
    axs[1].set_xlabel('Time [ms]', fontsize=12)
    axs[1].set_ylabel('Ground truth', fontsize=12)
    
    plt.tight_layout()


draw_pred_plot(18248, time_span=(1030, 1100))
```
